```python
import pandas as pd

df = pd.read_csv("real_estate_ads.csv", nrows=1000)

print(df.dtypes)
print(f"Total columns: {len(df.columns)}")
```

```python
for idx, (col, dtype) in enumerate(df.dtypes.items(), start=1):
    print(f"{idx} = {col}, {dtype}")
```

```python
type_mapping = {
    "object": "TEXT",
    "float64": "DOUBLE PRECISION",
    "int64": "BIGINT",
    "bool": "BOOLEAN"
}

columns = [
    f'"{col}" {type_mapping[str(dtype)]}' 
    for col, dtype in df.dtypes.items()     
]

sql = (
    "CREATE TABLE raw_listings (\n    "
    + ",\n    ".join(columns)
    + "\n);"
)

print(sql)
```
